```python
precio_cafe = 35.0
precio_sandwich = 45.0
precio_jugo = 25.0
total_dia = 0.0

while True:
    print("--- MENU DE SABOR TEC ---")
    print("1. Registrar pedido")
    print("2. Resumen del dia")
    print("3. Salir")
    opcion = int(input("Selecciona una opcion: "))

    if opcion == 1:
        nombre = input("Nombre del cliente: ")
        es_estudiante = input("Es estudiante (si/no): ").lower() == "si"
        c_cafe = int(input("Cantidad de cafes: "))
        c_sandwich = int(input("Cantidad de sandwiches: "))
        c_jugo = int(input("Cantidad de jugos: "))

        subtotal = (c_cafe * precio_cafe) + (c_sandwich * precio_sandwich) + (c_jugo * precio_jugo)
        descuento = 0.0
        if es_estudiante and subtotal >= 150:
            descuento = subtotal * 0.10
        iva = (subtotal - descuento) * 0.16
        total = subtotal - descuento + iva
        total_dia = total_dia + total

        print(f"Cliente: {nombre}")
        print(f"Subtotal: ${subtotal:.2f}")
        print(f"Descuento: ${descuento:.2f}")
        print(f"IVA (16%): ${iva:.2f}")
        print(f"Total: ${total:.2f}")

    elif opcion == 2:
        print(f"Ventas acumuladas del dia: ${total_dia:.2f}")

    elif opcion == 3:
        print("Cerrando el sistema del dia.")
        break
    else:
        print("Opcion invalida.")
```

```python
# ==============================================================================
# PROTOTIPO INICIAL - SISTEMA DE COBRO CARL'S JR.
# Curso: Solución de problemas con programación computacional
# Semana 3: Avance del Reto/Proyecto (Fase I)
# ==============================================================================

def mostrar_menu():
    print("\n" + "=" * 45)
    print("      SISTEMA DE CAJA - CARL'S JR.       ")
    print("=" * 45)
    print("1. Famous Star con Queso         $110.00")
    print("2. Western Bacon Cheeseburger    $135.00")
    print("3. Guacamole Bacon Burger        $150.00")
    print("4. Combo Super Star (Papa+Bebida)$180.00")
    print("5. Finalizar pedido y cobrar")
    print("-" * 45)

def ejecutar_caja():
    subtotal = 0.0
    continuar = "S"
    
    print("¡Bienvenido al sistema de punto de venta de Carl's Jr.!")
    
    while continuar.upper() == "S":
        mostrar_menu()
        try:
            opcion = int(input("Seleccione una opción (1-5): "))
        except ValueError:
            print("Error: Por favor ingrese un número entero válido.")
            continue
            
        if opcion >= 1 and opcion <= 4:
            try:
                cantidad = int(input("Ingrese la cantidad de piezas: "))
            except ValueError:
                print("Error: Cantidad inválida.")
                continue
                
            if cantidad > 0:
                if opcion == 1:
                    precio = 110.00
                    nombre_prod = "Famous Star con Queso"
                elif opcion == 2:
                    precio = 135.00
                    nombre_prod = "Western Bacon Cheeseburger"
                elif opcion == 3:
                    precio = 150.00
                    nombre_prod = "Guacamole Bacon Burger"
                elif opcion == 4:
                    precio = 180.00
                    nombre_prod = "Combo Super Star"
                    
                subtotal_item = precio * cantidad
                subtotal += subtotal_item
                print(f"-> Agregado: {cantidad}x {nombre_prod} = ${subtotal_item:.2f}")
                
                # Modificadores / Extras
                print("\n¿Desea agregar extras a este producto?")
                print("1. Tocino extra (+$20.00 c/u)")
                print("2. Queso extra (+$15.00 c/u)")
                print("3. Agrandar combo (+$25.00 c/u)")
                print("4. Ningún extra")
                
                try:
                    extra_opt = int(input("Seleccione una opción de extra (1-4): "))
                    if extra_opt == 1:
                        extra_costo = 20.00 * cantidad
                        subtotal += extra_costo
                        print("-> Se agregó Tocino extra.")
                    elif extra_opt == 2:
                        extra_costo = 15.00 * cantidad
                        subtotal += extra_costo
                        print("-> Se agregó Queso extra.")
                    elif extra_opt == 3:
                        extra_costo = 25.00 * cantidad
                        subtotal += extra_costo
                        print("-> Se agrande el combo.")
                except ValueError:
                    print("Opción de extra no válida. Se omite el extra.")
                    
                print(f"Subtotal acumulado de la orden: ${subtotal:.2f}")
            else:
                print("Error: La cantidad debe ser mayor a cero.")
                
        elif opcion == 5:
            continuar = "N"
        else:
            print("Opción no válida. Intente nuevamente (1-5).")
            
    if subtotal > 0:
        print("\n" + "=" * 45)
        tiene_vip = input("¿El cliente cuenta con Tarjeta VIP Carl's Jr.? (S/N): ")
        descuento = 0.0
        
        if tiene_vip.upper() == "S":
            descuento = subtotal * 0.15
            subtotal -= descuento
            print(f"¡Descuento VIP aplicado (15%): -${descuento:.2f}")
            
        iva = subtotal * 0.16
        total = subtotal + iva
        
        print("========================================")
        print("            TICKET DE VENTA             ")
        print("========================================")
        print(f"Subtotal con ajustes: ${subtotal:.2f}")
        print(f"IVA (16%):            ${iva:.2f}")
        print(f"TOTAL A PAGAR:        ${total:.2f}")
        print("========================================")
        print("¡Gracias por su visita a Carl's Jr.!")
        print("¡Disfrute su comida!")
    else:
        print("\nNo se registraron productos en la orden de caja.")

if __name__ == "__main__":
    ejecutar_caja()
```
