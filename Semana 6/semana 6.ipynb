```python
# Empezamos la suma en cero
suma = 0

# Revisamos los números del 1 al 20
for i in range(1, 21):
    # Si el número es impar, saltamos al siguiente
    if i % 2 != 0:
        continue
    suma = suma + i # Sumamos los pares

# Mostramos el resultado
print("La suma de los números pares del 1 al 20 es:", suma)
```

```python
# Función para sumar los números de una lista
def sumar_lista(lista):
    suma = 0
    for numero in lista:
        suma = suma + numero
    return suma # resultado

# Probamos
mi_lista = [10, 20, 30, 40]
resultado = sumar_lista(mi_lista)

print("La suma de los elementos es:", resultado)
```

```python
# encontramoa el número más grande
def encontrar_maximo(lista):
    maximo = lista[0] # Suponemos que el primero es el mayor
    for numero in lista:
        if numero > maximo:
            maximo = numero # Cambiamos el mayor si encontramos uno más grande
    return maximo

# Probamos la función
mi_lista = [7, 3, 9, 2]
resultado = encontrar_maximo(mi_lista)

print("El elemento máximo es:", resultado)
```

```python

def calcular_promedio(lista):
    suma = 0
    for numero in lista:
        suma = suma + numero
    
    promedio = suma / len(lista)
    print(f"El promedio es: {promedio:.2f}")
mi_lista = [8, 9, 10, 7]
calcular_promedio(mi_lista)
```

```python
# Función para calcular
def factorial(n):
    # base
    if n == 0:
        return 1
    else:
        # Se llama a sí misma
        return n * factorial(n - 1)

# Probamos la función
numero = 5
resultado = factorial(numero)

print("El factorial de", numero, "es:", resultado)
```

```python
# sumar los elementos de la tupla
def sumar_tupla(tupla):
    suma = 0
    for numero in tupla:
        suma = suma + numero
    return suma

# Creamos la tupla
mi_tupla = (4, 8, 15, 16, 23, 42)

# Imprimimos el tercer elemento (índice 2)
print("El tercer elemento es:", mi_tupla[2])

# Calculamos y mostramos la suma
resultado = sumar_tupla(mi_tupla)
print("La suma de la tupla es:", resultado)
```

```python
# Diccionario con las calificaciones
calificaciones = {"Ana": [8, 9, 10], "Beto": [7, 7, 8]}

# Revisamos cada alumno 
for nombre in calificaciones:
    notas = calificaciones[nombre]
    suma = 0
    
    # Sumamos
    for nota in notas:
        suma = suma + nota
    
    # Calculamos el promedio y lo mostramos
    promedio = suma / len(notas)
    print(f"{nombre}: {promedio:.2f}")
```
