```python
while True:
    print("\nMENU PRINCIPAL")
    print("1. Tuplas")
    print("2. Diccionarios")
    print("3. Excepciones")
    print("4. Strings")
    print("5. Salir")
    
    opcion = input("Elige una opcion (1-5): ")
    
    if opcion == "1":
        print("\n--- SECCION TUPLAS ---")
        numeros = (5, 12, 8, 20, 3)
        print("Tupla inicial:", numeros)
        print("Tercer elemento:", numeros[2])
        
        n1 = int(input("Ingresa un numero adicional: "))
        n2 = int(input("Ingresa otro numero adicional: "))
        nueva_tupla = numeros + (n1, n2)
        print("Nueva tupla:", nueva_tupla)
        
        lista = list(nueva_tupla)
        lista.sort()
        print("Tupla convertida a lista y ordenada:", lista)
        
        suma = 0
        for n in nueva_tupla:
            suma = suma + n
        print("Suma de todos los elementos:", suma)
        
    elif opcion == "2":
        print("\n--- SECCION DICCIONARIOS ---")
        contactos = {
            "Ana": "555-1111",
            "Luis": "555-2222",
            "Carmen": "555-3333"
        }
        print("Contactos iniciales:", contactos)
        
        nuevo_nombre = input("Ingresa el nombre del nuevo contacto: ")
        nuevo_tel = input("Ingresa el telefono: ")
        contactos[nuevo_nombre] = nuevo_tel
        
        print("Nombres registrados:")
        for nombre in contactos:
            print("-", nombre)
            
        buscar = input("Ingresa el nombre a buscar: ")
        if buscar in contactos:
            print("El telefono es:", contactos[buscar])
        else:
            print("El contacto no existe.")
            
    elif opcion == "3":
        print("\n--- SECCION EXCEPCIONES ---")
        try:
            num1 = int(input("Ingresa el primer numero entero: "))
            num2 = int(input("Ingresa el segundo numero entero: "))
            
            suma = num1 + num2
            print("La suma es:", suma)
            
            division = num1 / num2
            print("La division es:", division)
            
        except ValueError:
            print("Error: Debes ingresar numeros enteros validos.")
        except ZeroDivisionError:
            print("Error: No se puede dividir entre cero.")
            
    elif opcion == "4":
        print("\n--- SECCION STRINGS ---")
        mensaje = "aprender python es muy facil y divertido"
        print("Mensaje original:", mensaje)
        print("Longitud del mensaje:", len(mensaje))
        print("En mayusculas:", mensaje.upper())
        
        mensaje_cambiado = mensaje.replace("facil", "genial")
        print("Mensaje modificado:", mensaje_cambiado)
        
        palabras = mensaje.split()
        print("Cantidad de palabras:", len(palabras))
        
    elif opcion == "5":
        print("Programa finalizado. Adios.")
        break
    else:
        print("Opcion no valida, intenta de nuevo.")
```

```python
def sumar_calificaciones(tup):
    suma = 0
    for n in tup:
        suma = suma + n
    return suma

calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)
print("Tercera calificación:", calificaciones[2])

n1 = float(input("Nueva calificación 1: "))
n2 = float(input("Nueva calificación 2: "))

nueva_tupla = calificaciones + (n1, n2)
print("Nueva tupla:", nueva_tupla)

lista = list(nueva_tupla)
lista.sort(reverse=True)
print("Lista ordenada (mayor a menor):", lista)

suma_total = sumar_calificaciones(nueva_tupla)
print("Suma total:", suma_total)
```

```python
def buscar_telefono(agenda, nombre):
    return agenda[nombre]

agenda = {
    "Ana": "555-0101",
    "Luis": "555-0102",
    "Mía": "555-0103"
}

nuevo_nombre = input("Nombre del nuevo contacto: ")
nuevo_telefono = input("Teléfono del nuevo contacto: ")

agenda[nuevo_nombre] = nuevo_telefono

print("Contactos registrados:", ", ".join(agenda.keys()))

nombre_a_buscar = input("Nombre a buscar: ")
telefono_encontrado = buscar_telefono(agenda, nombre_a_buscar)

print(f"El teléfono de {nombre_a_buscar} es: {telefono_encontrado}")
```

```python
votos = {"Rojo": 0, "Azul": 0, "Verde": 0}

for _ in range(5):
    color = input("Color de su voto: ")
    if color in votos:
        votos[color] += 1

ganador = max(votos, key=votos.get)
print(f"Color ganador: {ganador}, Total: {votos[ganador]}")
```

```python
def contar_palabras(texto):
    return len(texto.split())

def buscar_telefono(agenda, nombre):
    return agenda.get(nombre, "No encontrado")

numeros = (45, 12, 89, 3, 27)
contactos = {"Ana": "555-0101", "Luis": "555-0102", "Mía": "555-0103"}

while True:
    print("\n--- Menú Principal ---")
    print("1. Mostrar tupla ordenada")
    print("2. Buscar teléfono")
    print("3. Dividir dos números")
    print("4. Analizar texto")
    print("5. Salir")
    
    opcion = input("Elige una opción: ")
    
    if opcion == "1":
        print("Tupla ordenada:", tuple(sorted(numeros)))
    elif opcion == "2":
        nombre = input("Nombre a buscar: ")
        print("Teléfono:", buscar_telefono(contactos, nombre))
    elif opcion == "3":
        try:
            num1 = float(input("Ingresa el primer número: "))
            num2 = float(input("Ingresa el segundo número: "))
            print("Resultado:", num1 / num2)
        except ZeroDivisionError:
            print("Error: No se puede dividir entre cero.")
        except ValueError:
            print("Error: Ingresa valores numéricos válidos.")
    elif opcion == "4":
        frase = input("Ingresa una frase: ")
        print("Longitud:", len(frase))
        print("En mayúsculas:", frase.upper())
        print("Texto reemplazado:", frase.replace("Python", "programación"))
        print("Palabras totales:", contar_palabras(frase))
    elif opcion == "5":
        break
    else:
        print("Opción inválida, intenta de nuevo.")
```
