# Ejercicio 1 Museo

```python
import pdb

print("=== COBRO DE ENTRADAS - MUSEO DE ANTROPOLOGIA E HISTORIA ===")

personas_a_ingresar = int(input("Ingrese la cantidad total de visitantes: "))

total_personas = 0
total_dinero = 0.0

contador = 0

while True:
    if contador == personas_a_ingresar:
        print("Se ha alcanzado el numero total de personas a ingresar.")
        break
        
    print("\n--- Registrar Visitante " + str(contador + 1) + " ---")
    
    # pdb.set_trace()
    
    edad_texto = input("Ingrese la edad (o escriba 'salir' para cancelar): ")
    
    if edad_texto == "salir" or edad_texto == "Salir":
        print("Saliendo del programa...")
        break
        
    if edad_texto == "":
        print("No ingresaste ninguna edad. Intenta de nuevo.")
        continue
        
    edad = int(edad_texto)
    
    if edad < 0 or edad > 120:
        print("Edad no valida. Intente de nuevo.")
        continue
    
    if edad < 3:
        precio = 0
        tipo = "Niño menor de 3 años"
    elif edad <= 17:
        precio = 30
        tipo = "Menor de edad"
    else:
        precio = 45
        tipo = "Mayor de edad"
        
    descuento = 0.0
    
    if precio > 0:
        print("Marque con 'si' las opciones que apliquen al visitante:")
        es_estudiante = input("¿Es estudiante? (si/no): ") == "si"
        es_profesor = input("¿Es profesor? (si/no): ") == "si"
        es_adulto_mayor = input("¿Es adulto mayor? (si/no): ") == "si"
        
        if es_adulto_mayor and not es_profesor and not es_estudiante:
            descuento = 0.12
            print("Descuento aplicado: Adulto Mayor (12%)")
        elif es_profesor and not es_adulto_mayor and not es_estudiante:
            descuento = 0.10
            print("Descuento aplicado: Profesor (10%)")
        elif es_estudiante and not es_adulto_mayor and not es_profesor:
            descuento = 0.10
            print("Descuento aplicado: Estudiante (10%)")
        elif es_adulto_mayor or es_profesor or es_estudiante:
            if es_adulto_mayor:
                descuento = 0.12
                print("Se aplicó el mayor descuento: Adulto Mayor (12%)")
            else:
                descuento = 0.10
                print("Se aplicó el descuento del 10%")
        else:
            print("Sin descuento aplicado.")
    
    monto_descuento = precio * descuento
    precio_final = precio - monto_descuento
    
    print("Tipo: " + tipo)
    print("Precio base: $" + str(precio) + " | Descuento: $" + str(monto_descuento) + " | Paga: $" + str(precio_final))
    
    total_personas += 1
    total_dinero += precio_final
    contador += 1

print("\n================ TOTAL DETALLADO ================")
lista_totales = [total_personas, total_dinero]

for x in lista_totales:
    print("Total acumulado: " + str(x))

print("=================================================")
```

```python
total_acumulado = 0.0

while True:
    edad = int(input("Ingrese la edad del visitante: "))
    
    if edad < 3:
        precio = 0
    elif edad <= 17:
        precio = 30
    else:
        precio = 45
        
    if precio == 0:
        print("Menor de 3 años, sin cargo")
        continue
        
    total_acumulado += precio
    
    if total_acumulado >= 100:
        print("Aforo alcanzado, se detiene el registro")
        break

print("Total acumulado: $" + str(total_acumulado))
```

# Ejercicio 3

```python
num_visitantes = int(input("Número de visitantes: "))

suma_edades = 0
adultos = 0

for i in range(num_visitantes):
    edad = int(input("Edad del visitante " + str(i + 1) + ": "))
    suma_edades += edad
    
    if edad >= 18:
        adultos += 1

promedio = suma_edades / num_visitantes

print("Adultos: " + str(adultos))
print("Promedio de edad: " + str(round(promedio, 2)))
```
