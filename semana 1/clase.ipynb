```python
print(" Calculadora de uso de aplicacciones")

nombre = input("Cual es tu nombre?")

instagram = float(input("Horas que se usan Instagram"))
Tik_tok = float(input("Horas que se usan Tik tok"))
Youtube = float(input("Horas que se usan Youtube"))
videojuegos = float(input("Horas que se usan videojuegos"))
facebook = float(input("Horas que se usan Facebook"))

tiempo_total= instagram + Tik_tok + Youtube + videojuegos + facebook

porcentaje_redes = (tiempo_total / 24) * 100

print(f"\n--- Resultado ---")
print(f"usuario: {nombre}")
print(f"Tiempo total de uso de redes: {tiempo_total:.2f} horas")
print(f"Porcentaje de tiempo en redes: {porcentaje_redes:.2f}%")
```

Ejercicio extra #1

```python
cuenta =  float(input("total de la cuenta: "))
propinas_porcentaje = float(input("Porcentaje de propina: "))
persona = int(input("numero de personas:"))

propina = cuenta + (cuenta * propinas_porcentaje / 100)
total = cuenta + propina
pago_por_persona = total / persona

print(f"--- Resultado ---")
print(f"total de la cuenta: ${cuenta:.2f}")
print(f"propina: ${propina:.2f}")
print(f"total a pagar: ${total:.2f}")
print(f"pago por persona: ${pago_por_persona:.2f}")
```
