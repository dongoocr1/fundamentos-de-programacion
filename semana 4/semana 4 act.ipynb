```python
tabla = []

for i in range(1, 11):
    fila = []
    for j in range(1, 11):
        num = 0
        for k in range(j):
            num = num + i
        fila.append(num)
    tabla.append(fila)

def mostrar_tabla(matriz):
    for r in matriz:
        for c in r:
            print(c, end="\t")
        print()

def obtener_multiplicacion(matriz, factor1, factor2):
    res = matriz[factor1 - 1][factor2 - 1]
    return res

mostrar_tabla(tabla)

f1 = int(input("Ingresa el primer factor (1-10): "))
f2 = int(input("Ingresa el segundo factor (1-10): "))

resultado = obtener_multiplicacion(tabla, f1, f2)
print("El resultado", f1, "x", f2, "es:", resultado)
```

```python
def imprimir_matriz(matriz):
    for fila in matriz:
        for elemento in fila:
            print(elemento, end=" ")
        print()

def suma_matriz(matriz):
    suma = 0
    for fila in matriz:
        for elemento in fila:
            suma = suma + elemento
    return suma

m = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

imprimir_matriz(m)

total = suma_matriz(m)
print("La suma de la matriz es:", total)
```

```python
def suma_fila(fila):
    suma = 0
    for num in fila:
        suma = suma + num
    return suma

def mostrar_tabla(matriz):
    for f in matriz:
        for num in f:
            print(num, end=" ")
        print()

m = [[3, 1, 4], [1, 5, 9], [2, 6, 5]]

mostrar_tabla(m)

for i in range(len(m)):
    res = suma_fila(m[i])
    print("Suma de la fila", i, ":", res)
```

```python
def multiplicar_columna(matriz, columna):
    prod = 1
    for fila in matriz:
        num = fila[columna]
        temp = 0
        for i in range(num):
            temp = temp + prod
        prod = temp
    return prod

m = [[1, 2, 3], [4, 5, 6], [7, 8, 9], [2, 4, 6]]

col = int(input("Numero de columna: "))

res = multiplicar_columna(m, col)
print("El producto de la columna", col, "es:", res)
```

```python
def generar_tabla(tamaño):
    matriz = []
    for i in range(1, tamaño + 1):
        fila = []
        for j in range(1, tamaño + 1):
            num = 0
            for k in range(j):
                num = num + i
            fila.append(num)
        matriz.append(fila)
    return matriz

def imprimir_tabla(tabla):
    for f in tabla:
        for c in f:
            print(c, end="\t")
        print()

n = int(input("Tamaño de la tabla: "))

if n >= 2 and n <= 5:
    t = generar_tabla(n)
    imprimir_tabla(t)
    
    r = int(input("Renglon (factor): "))
    c = int(input("Columna (factor): "))
    
    if r >= 1 and r <= n and c >= 1 and c <= n:
        res = t[r - 1][c - 1]
        print("El producto de", r, "x", c, "es:", res)
else:
    print("El tamaño debe estar entre 2 y 5")
```
